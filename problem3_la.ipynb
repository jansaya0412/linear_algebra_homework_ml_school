```python
import numpy as np

rng = np.random.default_rng(0)
```

```python
def rref(A, tol=1e-10):
    """Return (R, pivots): the RREF of A and the list of pivot columns."""
    R = np.array(A, dtype=float)
    m, n = R.shape #m is number of rows  and n is number of columns

    pivots = []
    r = 0 #to keep track of the row where the next pivot is places

    for c in range(n):
        if r == m: #if all rows contain pivots
            break

        pivot_row = r 

        for i in range(r + 1, m): #to find the entry with the largest absolute value
            if abs(R[i, c]) > abs(R[pivot_row, c]):
                pivot_row = i
        
        #if it is smaller than tol,the column has no pivot, so we skip it
        if abs(R[pivot_row, c]) < tol:
            continue
    
        #otherwise, swaps its row with row r
        R[[r, pivot_row]] = R[[pivot_row, r]]
        
        #divide row r by the pivot
        pivot = R[r, c]
        R[r] = R[r] / pivot
        
        #subtracts multiples of row r from all the other rows
        for i in range(m):
            if i != r:
                factor = R[i, c]
                R[i] = R[i] - factor * R[r]

        pivots.append(c)
        r += 1
    
    #replaces very small values with zero
    R[abs(R) < tol] = 0

    return R, pivots
```

```python
A = np.array([
    [1, 2, -1],
    [3, -1, 4]])

R, pivots = rref(A)
R
```

```python
def rank(A):
    R, pivots = rref(A)
    rank = len(pivots)
    return rank
```

```python
print(rank(A))
```

```python
def col_space_basis(A):
    R, pivots = rref(A)
    return pivots

print(col_space_basis(A))
```

```python
def null_space_basis(A):
    R, pivots = rref(A)
    n = R.shape[1]
    basis = []

    for j in range(n):
        if j not in pivots:
            v = np.zeros(n)
            v[j] = 1

            for i in range(len(pivots)):
                v[pivots[i]] = -R[i, j]

            basis.append(v)

    return basis

print(null_space_basis(A))
```

```python
for k in range(20):
    A = rng.normal(size=(5, 7))
    basis = null_space_basis(A)

    for v in basis:
        if not np.allclose(A @ v, 0):
            print(k + 1, "not close to 0")
        else:
            print(k + 1, "close to 0")
            
    if rank(A) + len(basis) != 7:
        print(k + 1, "rank-nullity not passed")
    else:
        print(k + 1, "rank-nullity passed")

    if rank(A) != np.linalg.matrix_rank(A):
        print(k + 1, "rank doesnt agree with np.linalg.matrix_rank(A)")
    else:
        print(k + 1, "rank agrees")
```

```python
B = rng.normal(size=(6, 4))
C = rng.normal(size=(4, 8))

A = B @ C

rank = rank(A)
basis = null_space_basis(A)

if rank == 4:
    print("found rank 4")
else:
    print("rank 4 not found")

if len(basis) == 4:
    print("4-dimensional null space")
else:
    print("not 4-dimensional")
```
